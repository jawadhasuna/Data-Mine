```python
#matrix mul
import numpy as np

# X: m samples x n features
X = np.array([[0,0],[0,1],[1,0],[1,1]])
t = np.array([0, 0, 0, 1])  # AND logic

# Add bias term
X_aug = np.hstack([X, np.ones((X.shape[0], 1))])  # shape: (m, n+1)
w = np.array([1, 1, -1.5])  # weights + bias

# Compute perceptron output for all samples
y = (X_aug @ w > 0).astype(int)
print(y)  # [0 0 0 1]
```

```python
#perceptron learning
import numpy as np

def perceptron_train(X, t, eta=0.1, epochs=10):
    n_samples, n_features = X.shape
    w = np.zeros(n_features)
    b = 0
    
    for epoch in range(epochs):
        for xi, ti in zip(X, t):
            a = np.dot(w, xi) + b
            y = 1 if a > 0 else 0
            w += eta * (ti - y) * xi
            b += eta * (ti - y)
    return w, b
```
